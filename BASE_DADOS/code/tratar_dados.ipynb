```python
# 1. Método de contagem de Nulos/NaNs nativos do Pandas
null_counts = df.isnull().sum()


# Define o limite máximo de linhas exibidas pelo pandas
pd.set_option("display.max_rows", 200)

# (Opcional) Define o limite de colunas para não cortar colunas na largura
pd.set_option("display.max_columns", 120)

# 1. Impede o Pandas de quebrar colunas para a linha de baixo
pd.set_option('display.expand_frame_repr', False)

# 2. Aumenta o limite de largura total da linha (em caracteres)
pd.set_option('display.width', 1000)

# 3. Exibe todas as colunas da tabela na mesma linha
pd.set_option('display.max_columns', None)


# 2. Aplicação do método
empty_str_counts = {}
for col in df.columns:
    if df[col].dtype == "object":
        cleaned = df[col].astype(str).str.strip().str.lower()
        empty_mask = cleaned.isin(["", "nan", "null", "none", ".", "-", "n/a"])
        empty_str_counts[col] = empty_mask.sum()
    else:
        empty_str_counts[col] = 0

# 3. Consolidação do diagnóstico
df_diagnostico = pd.DataFrame(
    {
        "coluna": df.columns,
        "nulos_nativos": null_counts.values,
        "textos_vazios": [empty_str_counts[c] for c in df.columns],
    }
)

print(df_diagnostico)
```

```python
#Conversão dos tipos
# 1. Definir identificadores da estrutura de painel
col_data = "mes_ano"
col_painel = "porte"

# 2. Converter mes_ano para datetime
df[col_data] = pd.to_datetime(df[col_data], errors="coerce") 


print(df.head(14))
```

```python
# 3. Converter todas as variáveis numéricas para float64
for col in df.columns:
    if col not in [col_data, col_painel]:
        if df[col].dtype == "object":
            # Trata padrão brasileiro de pontuação (remove ponto de milhar e troca vírgula por ponto)
            df[col] = (
                df[col]
                .astype(str)
                .str.replace(".", "", regex=False)
                .str.replace(",", ".", regex=False)
                .str.strip()
            )
            df[col] = pd.to_numeric(df[col], errors="coerce")
        elif df[col].dtype == "int64":
            df[col] = df[col].astype("float64")

print(df.head(14))
```

```python
# Nova verificação dos nulos
# 1. Método de contagem de Nulos/NaNs nativos do Pandas
null_counts = df.isnull().sum()


# Define o limite máximo de linhas exibidas pelo pandas
pd.set_option("display.max_rows", 200)

# (Opcional) Define o limite de colunas para não cortar colunas na largura
pd.set_option("display.max_columns", 120)

# 1. Impede o Pandas de quebrar colunas para a linha de baixo
pd.set_option('display.expand_frame_repr', False)

# 2. Aumenta o limite de largura total da linha (em caracteres)
pd.set_option('display.width', 1000)

# 3. Exibe todas as colunas da tabela na mesma linha
pd.set_option('display.max_columns', None)


# 2. Aplicação do método
empty_str_counts = {}
for col in df.columns:
    if df[col].dtype == "object":
        cleaned = df[col].astype(str).str.strip().str.lower()
        empty_mask = cleaned.isin(["", "nan", "null", "none", ".", "-", "n/a"])
        empty_str_counts[col] = empty_mask.sum()
    else:
        empty_str_counts[col] = 0

# 3. Consolidação do diagnóstico
df_diagnostico = pd.DataFrame(
    {
        "coluna": df.columns,
        "nulos_nativos": null_counts.values,
        "textos_vazios": [empty_str_counts[c] for c in df.columns],
    }
)

print(df_diagnostico)
```

```python
# 1. Caminho do arquivo
caminho_arquivo = r"C:\Users\joaov\OneDrive\Tudo\Estudos\MBA-USP\TCC\BASE_DADOS\data\master_tcc_final_consolidated.csv"

df = pd.read_csv(caminho_arquivo, sep=';', encoding='utf-8-sig')

# 2. Remoção de espaços em branco nas colunas categóricas
df['porte'] = df['porte'].str.strip()

# 3. Chaves e Dummies (NÃO devem ser testadas no ADF nem tratadas como float contínuo)
colunas_chave = ['mes_ano', 'porte']
colunas_dummy = ['is_pandemic', 'is_lockdown']

# Identifica apenas as colunas CONTÍNUAS verdadeiras
num_cols = [col for col in df.columns if col not in colunas_chave + colunas_dummy]


for col in num_cols:
    df[col] = df[col].astype(str).str.replace(',', '.').astype(float)


def parse_mes_ano(val):
    month_str, year_str = val.split('/')
    m = int(month_str)
    y = int(year_str)
    return pd.Timestamp(year=y, month=m, day=1)

df['data'] = df['mes_ano'].apply(parse_mes_ano)

# 5. Ordenação temporal e reordenação das colunas
cols_order = ['data', 'mes_ano', 'porte'] + colunas_dummy + num_cols
df = df.sort_values(by=['data', 'porte']).reset_index(drop=True)
df = df[cols_order]

display(cols_order)
display(df['is_pandemic'])
display(df)
```

```python
target_col = 'default_percent' 
regressors_cols = [c for c in num_cols if c != target_col]

print("\n--- Executando Teste ADF de Estacionariedade na Série Agregada ---")

adf_results = []
for col in [target_col] + regressors_cols:
    # Executa ADF na série sem valores nulos
    series_clean = df[col].dropna()
    res = adfuller(series_clean, autolag='AIC')
    
    p_value = res[1]
    is_stationary = "Sim (I(0))" if p_value < 0.05 else "Não (I(1)) - Necessita Diferença"
    
    adf_results.append({
        'Variável': col,
        'Estatística ADF': round(res[0], 4),
        'p-valor': round(p_value, 4),
        'Estacionária?': is_stationary
    })

df_adf_summary = pd.DataFrame(adf_results)
display(df_adf_summary.head(15)) # Exibe as primeiras variáveis para avaliação
```

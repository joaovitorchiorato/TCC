```python
import pandas as pd

# 1. Carregar o arquivo Parquet preservando todos os dtypes
df = pd.read_parquet('../data/dataset_macro_painel_limpo_com_dummies.parquet')

# 2. Conferir os tipos de dados, incluindo os dois indicadores binários
print("=== CONTAGEM DOS TIPOS CARREGADOS ===")
print(df.dtypes.value_counts())

# 3. Visualizar as primeiras linhas
df.head()
```

```python
import pandas as pd

# 1. Ordenar cronologicamente por porte e mes_ano
df = df.sort_values(by=["porte", "mes_ano"]).reset_index(drop=True)

# 2. Definir a estrutura de MultiIndex para Painel (Entidade: porte, Tempo: mes_ano)
df_panel = df.set_index(["porte", "mes_ano"])

# 3. Verificar a quantidade de observações por porte
obs_por_porte = df_panel.groupby(level="porte").size()

print("=== NÚMERO DE OBSERVAÇÕES POR PORTE ===")
print(obs_por_porte)

print("\n=== ESTRUTURA E BALANCEAMENTO ===")
print(f"Total de portes (entidades): {df_panel.index.levels[0].nunique()}")
print(f"Período coberto: {df_panel.index.levels[1].min().strftime('%m/%Y')} a {df_panel.index.levels[1].max().strftime('%m/%Y')}")

e_balanceado = (obs_por_porte.nunique() == 1) and (obs_por_porte.iloc[0] == df_panel.index.levels[1].nunique())
print(f"O painel é estritamente balanceado? {e_balanceado}")
```

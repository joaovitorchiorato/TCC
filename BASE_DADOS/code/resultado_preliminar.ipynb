```python
data_folder2 = '../data'
file_name2 = 'master_tcc_final_consolidated.csv'
file_path2 = os.path.join(data_folder2, file_name2)

df_teste2 = pd.read_csv(file_path2, sep=';', encoding='utf-8-sig')
display(df_teste2)


df_tcc_final = df_teste2


df_tcc_final = df_tcc_final.drop(columns=['vencido_acima_de_15_dias', 'carteira_inadimplida_arrastada',	'ativo_problematico',	'carteira_ativa'])

print('pib_12m_dolar2' in df_tcc_final.columns)

display(df_tcc_final)
```

```python
# 1. Converte a coluna para o formato de data real do Pandas
# (Substitua 'mes_ano' pelo nome exato da sua coluna de data)
df_tcc_final['mes_ano'] = pd.to_datetime(df_tcc_final['mes_ano'], format='%m/%Y')

# 2. Define essa coluna de data como o índice do seu DataFrame
df_tcc_final.set_index('mes_ano', inplace=True)

# 3. Extrai o número do trimestre de cada linha (valores de 1 a 4)
trimestre_num = df_tcc_final.index.quarter

# 4. Cria as 3 dummies de trimestre automaticamente
df_tcc_final['trimestre_1'] = (trimestre_num == 1).astype(int)
df_tcc_final['trimestre_2'] = (trimestre_num == 2).astype(int)
df_tcc_final['trimestre_3'] = (trimestre_num == 3).astype(int)

display(df_tcc_final)
```

```python
# 1. Transforma a coluna de porte em dummies, omitindo a primeira delas (drop_first=True)
# Isso vai criar automaticamente as 12 dummies necessárias
df_porte_dummies = pd.get_dummies(df_tcc_final['porte'], prefix='porte', drop_first=True, dtype=int)

# 2. Junta as novas colunas de porte de volta no seu DataFrame principal
df_completo = pd.concat([df_tcc_final, df_porte_dummies], axis=1)

# 3. Agora você já pode deletar a coluna de texto original do porte para não confundir o modelo
df_completo.drop(columns=['porte'], inplace=True)
```

```python
# 1. Cria uma cópia limpa para resolver a fragmentação de memória (PerformanceWarning)
df_final_limpo = df_completo.copy()

colunas_dummies = [
    # Sazonalidade Trimestral
    'trimestre_1', 
    'trimestre_2', 
    'trimestre_3', 
    
    # Choques da Pandemia
    'is_pandemic', 
    'is_lockdown', 
    
    # Dummies de Porte (Pessoa Física e Pessoa Jurídica)
    'porte_PF - Até 1 salário mínimo                    ',
    'porte_PF - Indisponível                            ',
    'porte_PF - Mais de 1 a 2 salários mínimos          ',
    'porte_PF - Mais de 10 a 20 salários mínimos        ',
    'porte_PF - Mais de 2 a 3 salários mínimos          ',
    'porte_PF - Mais de 3 a 5 salários mínimos          ',
    'porte_PF - Mais de 5 a 10 salários mínimos         ',
    'porte_PF - Sem rendimento                          ',
    'porte_PJ - Grande                                  ',
    'porte_PJ - Indisponível                            ',
    'porte_PJ - Micro                                   ',
    'porte_PJ - Médio                                   ',
    'porte_PJ - Pequeno                                 '
]

# 2. Identifica apenas as colunas econômicas e seus lags (tudo o que NÃO está na lista de dummies)
colunas_para_tratar = [col for col in df_final_limpo.columns if col not in colunas_dummies]

# 3. Varre as colunas aplicando a regra de forma condicional (segura)
for col in colunas_para_tratar:
    # SÓ aplica a limpeza de texto se a coluna for do tipo 'object' (string)
    if df_final_limpo[col].dtype == 'object':
        df_final_limpo[col] = df_final_limpo[col].astype(str).str.replace(',', '.')
        df_final_limpo[col] = pd.to_numeric(df_final_limpo[col], errors='coerce')
    else:
        # Se a coluna já for numérica (float ou int), apenas garante o formato padrão
        df_final_limpo[col] = pd.to_numeric(df_final_limpo[col], errors='coerce')

# 4. Remove linhas com valores nulos que possam ter surgido
df_final_limpo.dropna(inplace=True)

# 5. Salva de volta na sua variável principal
df_completo = df_final_limpo.copy()
```

```python
colunas_dummies = [
    # Sazonalidade Trimestral
    'trimestre_1', 
    'trimestre_2', 
    'trimestre_3', 
    
    # Choques da Pandemia
    'is_pandemic', 
    'is_lockdown', 
    
    # Dummies de Porte (Pessoa Física e Pessoa Jurídica)
    'porte_PF - Até 1 salário mínimo                    ',
    'porte_PF - Indisponível                            ',
    'porte_PF - Mais de 1 a 2 salários mínimos          ',
    'porte_PF - Mais de 10 a 20 salários mínimos        ',
    'porte_PF - Mais de 2 a 3 salários mínimos          ',
    'porte_PF - Mais de 3 a 5 salários mínimos          ',
    'porte_PF - Mais de 5 a 10 salários mínimos         ',
    'porte_PF - Sem rendimento                          ',
    'porte_PJ - Grande                                  ',
    'porte_PJ - Indisponível                            ',
    'porte_PJ - Micro                                   ',
    'porte_PJ - Médio                                   ',
    'porte_PJ - Pequeno                                 '
]

# 2. Identifica as variáveis contínuas (tirando as dummies E o default_percent)
colunas_economicas = [col for col in df_completo.columns if col not in colunas_dummies and col != 'default_percent']

# 3. Loop automático para criar os lags de 1, 3 e 12 meses
for col in colunas_economicas:
    df_completo[f'{col}_lag1'] = df_completo[col].shift(1)
    df_completo[f'{col}_lag3'] = df_completo[col].shift(3)
    df_completo[f'{col}_lag12'] = df_completo[col].shift(12)

# 4. Limpar as linhas com valores vazios (NaN) gerados pelo shift nos primeiros meses
df_completo.dropna(inplace=True)
```

```python
# Gerar a matriz de correlação em valores absolutos
matriz_corr = df_completo.corr().abs()

# Filtrar para exibir apenas pares com correlação muito alta (acima de 0.85)
# Exclui a diagonal principal (correlação da variável com ela mesma)
correlacoes_altas = matriz_corr[matriz_corr > 0.85].stack().reset_index()
correlacoes_altas = correlacoes_altas[correlacoes_altas['level_0'] != correlacoes_altas['level_1']]

# Exibir os pares colineares para você analisar
print("Pares com alta multicolinearidade (Pearson > 0.85):")
print(correlacoes_altas)
```

```python
# 1. Calcular a matriz de correlação de Pearson pura para a base toda
matriz_corr_completa = df_completo.corr(numeric_only=True)

# 2. Isolar apenas a coluna da variável alvo (default_percent)
corr_alvo = matriz_corr_completa[['default_percent']].reset_index()
corr_alvo.columns = ['Variavel', 'Pearson_r']

# 3. FILTRO: Remover da lista tudo o que for LAG ou DUMMY
# Remove os lags (colunas que terminam com _lag1, _lag3, _lag12)
corr_alvo = corr_alvo[~corr_alvo['Variavel'].str.contains('_lag')]

# Remove as dummies da lista (porte, trimestre e pandemia)
corr_alvo = corr_alvo[~corr_alvo['Variavel'].isin(colunas_dummies)]

# Remove a própria variável alvo da lista (correlação dela com ela mesma é 1.0)
corr_alvo = corr_alvo[corr_alvo['Variavel'] != 'default_percent']

# 4. Ordenar do maior impacto (seja positivo ou negativo) para o menor
corr_alvo['Impacto_Absoluto'] = corr_alvo['Pearson_r'].abs()
corr_alvo = corr_alvo.sort_values(by='Impacto_Absoluto', ascending=False)
corr_alvo.drop(columns=['Impacto_Absoluto'], inplace=True)

# 5. Configurar o Pandas para exibir todas as linhas na tela sem resumir (...)
pd.set_option('display.max_rows', 200)
pd.set_option('display.max_colwidth', None)

# 6. Exibir o resultado final pronto para copiar para o Word/LaTeX do TCC
print("Correlação de Pearson direta com a Inadimplência (Sem Lags e Sem Dummies):")
print(corr_alvo)

# 7. Salvar em CSV para facilitar o preenchimento da sua tabela
corr_alvo.to_csv('correlacao_direta_alvo_sem_lags.csv', index=False, sep=';', encoding='utf-8-sig')
```

```python
# 1. Isolar as colunas contínuas contemporâneas (exclui dummies e lags)
colunas_macro = [col for col in df_completo.columns if col not in colunas_dummies and '_lag' not in col and col != 'default_percent']

# 2. Calcular a correlação das explicativas entre si (X com X)
matriz_X = df_completo[colunas_macro].corr().abs()

# 3. Desempilhar para ver os pares
pares_colineares = matriz_X.stack().reset_index()
pares_colineares.columns = ['Variavel_A', 'Variavel_B', 'Pearson_r']

# 4. Filtros: remove autocorrelação e duplicidade espelhada (A-B e B-A)
pares_colineares = pares_colineares[pares_colineares['Variavel_A'] != pares_colineares['Variavel_B']]
pares_colineares['par_unico'] = pares_colineares.apply(lambda row: tuple(sorted([row['Variavel_A'], row['Variavel_B']])), axis=1)
pares_colineares.drop_duplicates(subset=['par_unico'], inplace=True)

# 5. Filtrar APENAS o bloco crítico de multicolinearidade (> 0.85)
alta_colinearidade = pares_colineares[pares_colineares['Pearson_r'] > 0.85].sort_values(by='Pearson_r', ascending=False)
alta_colinearidade.drop(columns=['par_unico'], inplace=True)

# Configuração para exibir tudo na tela do Jupyter
pd.set_option('display.max_rows', 150)
print(f"Total de pares colineares encontrados: {len(alta_colinearidade)}")
print(alta_colinearidade)
```

```python
# 1. Lista definitiva das 20 variáveis que você escolheu manter
variaveis_selecionadas = [
    'selic', 'ipca', 'unemployment', 'ibc_br', 'usd', 'igp_m', 
    'pib_12m_real', 'salario_min', 'ibovespa', 'endividamento_familias',
    'comprometimento_renda', 'indic_confianca_consumi', 'saldo_carteira_credito_total',
    'div_liq_set_publico_consolidado', 'papel_moeda_emitido', 'consum_deriv_tot',
    'consum_energ_ele_tot', 'emp_for_tot', 'exp_total_usd', 'imp_total_usd'
]

# 2. Garantir que a variável alvo e as dummies estruturais fiquem na base
colunas_finais = ['default_percent'] + [col for col in colunas_dummies if col in df_completo.columns]
df_tcc_final = df_completo[colunas_finais].copy()

# 3. Gerar os lags (1, 3 e 12) estritamente para as 20 selecionadas
for col in variaveis_selecionadas:
    if col in df_completo.columns:
        # Mantém a contemporânea
        df_tcc_final[col] = df_completo[col]
        # Cria os lags
        df_tcc_final[f'{col}_lag1'] = df_completo[col].shift(1)
        df_tcc_final[f'{col}_lag3'] = df_completo[col].shift(3)
        df_tcc_final[f'{col}_lag12'] = df_completo[col].shift(12)

# 4. Remover as linhas com NaN que os lags criam no começo da série
df_tcc_final.dropna(inplace=True)

print("==================================================================")
print("BASE DE DADOS DO TCC CONFIGURADA COM SUCESSO!")
print(f"Total de linhas (meses): {df_tcc_final.shape[0]}")
print(f"Total de colunas (Preditoras + Lags + Dummies): {df_tcc_final.shape[1]}")
print("==================================================================")

# Salvar a base oficial do modelo
df_tcc_final.to_csv('base_modelagem_final_tcc.csv', sep=';', encoding='utf-8-sig')
```

```python
# 1. Isolar apenas as colunas econômicas contínuas contemporâneas
colunas_macro = [col for col in df_completo.columns if col not in colunas_dummies and '_lag' not in col and col != 'default_percent']

# 2. Calcular a correlação contemporânea de todo mundo contra o alvo (default_percent)
corr_alvo = df_completo[colunas_macro + ['default_percent']].corr(numeric_only=True)['default_percent']

# --- FILTRO 1: CORRELAÇÃO FRACA (|r| < 0.05) ---
# Identifica quem tem impacto linear contemporâneo quase nulo
variaveis_fortes_alvo = corr_alvo[corr_alvo.abs() >= 0.05].drop('default_percent', errors='ignore').index.tolist()
variaveis_eliminadas_fracas = [c for c in colunas_macro if c not in variaveis_fortes_alvo]

# --- FILTRO 2: MULTICOLINEARIDADE CRÍTICA (> 0.85) ---
# Calcular a matriz de correlação (X com X) apenas das que passaram no Filtro 1
matriz_X = df_completo[variaveis_fortes_alvo].corr().abs()

variaveis_eliminar_colinear = set()
colunas_varridas = list(matriz_X.columns)

for i in range(len(colunas_varridas)):
    for j in range(i + 1, len(colunas_varridas)):
        v1 = colunas_varridas[i]
        v2 = colunas_varridas[j]
        
        # Se elas forem muito colineares entre si
        if matriz_X.loc[v1, v2] > 0.85:
            # Critério de desempate de Pearson: elimina a que tiver menor correlação absoluta com o alvo
            if abs(corr_alvo[v1]) >= abs(corr_alvo[v2]):
                variaveis_eliminar_colinear.add(v2)
            else:
                variaveis_eliminar_colinear.add(v1)

# 3. Lista final das sobreviventes que passaram em ambos os filtros de Pearson
variaveis_mantidas_final = [c for c in variaveis_fortes_alvo if c not in variaveis_eliminar_colinear]

# --- RELATÓRIO DO SEU CRITÉRIO DE PEARSON ---
print("==================================================================")
print("TRIAGEM BASEADA ESTRITAMENTE EM PEARSON")
print("==================================================================")
print(f"Variáveis macroeconômicas iniciais: {len(colunas_macro)}")
print(f"1. Eliminadas por baixo impacto (|r| < 0.05): {len(variaveis_eliminadas_fracas)}")
print(f"2. Eliminadas por multicolinearidade (> 0.85): {len(variaveis_eliminar_colinear)}")
print(f"-> VARIÁVEIS SELECIONADAS PARA O MODELO: {len(variaveis_mantidas_final)}")
print("==================================================================\n")

# Exibir a tabela final com os valores reais de Pearson com o alvo
df_tabela_final = pd.DataFrame({'Pearson_com_Alvo': corr_alvo[variaveis_mantidas_final]})
df_tabela_final['Absoluto'] = df_tabela_final['Pearson_com_Alvo'].abs()
df_tabela_final = df_tabela_final.sort_values(by='Absoluto', ascending=False).drop(columns=['Absoluto'])

print("LISTA DE VARIÁVEIS SELECIONADAS PELO SEU CRITÉRIO DE PEARSON:")
print(df_tabela_final)
```

```python
# 1. Isolar TODAS as variáveis explicativas que possuem lags (exclui apenas dummies e o alvo)
# Aqui deixamos os lags entrarem na dança
colunas_com_lags = [col for col in df_completo.columns if col not in colunas_dummies and col != 'default_percent']

# 2. Calcular a correlação de absolutamente todo mundo contra o alvo (default_percent)
corr_alvo_lags = df_completo[colunas_com_lags + ['default_percent']].corr(numeric_only=True)['default_percent']

# --- FILTRO 1: CORRELAÇÃO FRACA COM O ALVO (|r| < 0.05) ---
variaveis_fortes_lags = corr_alvo_lags[corr_alvo_lags.abs() >= 0.05].drop('default_percent', errors='ignore').index.tolist()
eliminadas_fracas_lags = [c for c in colunas_com_lags if c not in variaveis_fortes_lags]

# --- FILTRO 2: MULTICOLINEARIDADE CRÍTICA ENTRE VARIÁVEIS E LAGS (> 0.85) ---
matriz_X_lags = df_completo[variaveis_fortes_lags].corr().abs()

variaveis_eliminar_lags = set()
colunas_varridas_lags = list(matriz_X_lags.columns)

for i in range(len(colunas_varridas_lags)):
    for j in range(i + 1, len(colunas_varridas_lags)):
        v1 = colunas_varridas_lags[i]
        v2 = colunas_varridas_lags[j]
        
        # Se houver colinearidade pesada (ex: selic_lag1 com selic_lag3 > 0.85)
        if matriz_X_lags.loc[v1, v2] > 0.85:
            # Desempate de Pearson: elimina o lag/variável que tiver menor correlação absoluta com o alvo
            if abs(corr_alvo_lags[v1]) >= abs(corr_alvo_lags[v2]):
                variaveis_eliminar_lags.add(v2)
            else:
                variaveis_eliminar_lags.add(v1)

# 3. Lista final das sobreviventes (Variáveis e Lags específicos)
features_finais_tcc = [c for c in variaveis_fortes_lags if c not in variaveis_eliminar_lags]

# --- RELATÓRIO DA TRIAGEM COMPLETA COM LAGS ---
print("==================================================================")
print("TRIAGEM DE PEARSON APLICADA DIRETAMENTE NOS LAGS")
print("==================================================================")
print(f"Total de colunas analisadas (Variáveis + Lags): {len(colunas_com_lags)}")
print(f"1. Eliminadas por baixo impacto com o alvo (|r| < 0.05): {len(eliminadas_fracas_lags)}")
print(f"2. Eliminadas por multicolinearidade crítica (> 0.85): {len(variaveis_eliminar_lags)}")
print(f"-> VARIÁVEIS/LAGS SELECIONADOS PARA O MODELO: {len(features_finais_tcc)}")
print("==================================================================\n")

# Exibir a tabela final de sobreviventes
df_resultado_lags = pd.DataFrame({'Pearson_com_Alvo': corr_alvo_lags[features_finais_tcc]})
df_resultado_lags['Absoluto'] = df_resultado_lags['Pearson_com_Alvo'].abs()
df_resultado_lags = df_resultado_lags.sort_values(by='Absoluto', ascending=False).drop(columns=['Absoluto'])

print("LISTA DEFINITIVA DE COMPONENTES PARA O SEU MODELO:")
print(df_resultado_lags)

# 4. Salvar a base de dados final do TCC contendo apenas essas colunas sobreviventes + dummies + alvo
colunas_salvar = ['default_percent'] + [c for c in colunas_dummies if c in df_completo.columns] + features_finais_tcc
df_final_tcc_real = df_completo[colunas_salvar].dropna()
df_final_tcc_real.to_csv('base_tcc_filtrada_com_lags_pearson.csv', sep=';', encoding='utf-8-sig')
```
